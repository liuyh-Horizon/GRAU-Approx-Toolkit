# Import GRAU Approximation Toolkit

```python
import numpy as np
from GRAUApproxToolkit import GRAU_PWLF
```

# Test Piecewise Linearization Fitting of Quantized Sigmoid

### Prepare dummy dataset of quantized Sigmoid and create the piecewise linearization fitted Sigmoid function

```python
data_in  = np.random.uniform(-6, 6, 10000)
data_in  = np.sort(data_in)
data_out = 1 / (1 + np.exp(-data_in))

x  = np.round(data_in * 200)
y = np.round(data_out * 255)

pwlf = GRAU_PWLF(x, y, segment_num=6)  
```

### Plot the original PWLF Sigmoid and its PoT and APoT approximated functions

```python
pwlf.plot_final("original")
pwlf.plot_final("PoT")
pwlf.plot_final("APoT")
```

### Plot the PWLF steps for this Sigmoid function

```python
pwlf.plot_steps()
```

### Plot the comparison between original Sigmoid function and its PWLF, PoT-PWLF, and APoT-PWLF approxiamted functions

```python
pwlf.plot_compare("Sigmoid")
```

### Print all breaking points

```python
print(pwlf.breakpoints)
print(pwlf.segment_num)
```

### Print slopes of PWLF, PoT-PWLF, and APoT-PWLF approximated Sigmoid

```python
print(pwlf.slopes("original"))
print(pwlf.slopes("PoT"))
print(pwlf.slopes("PoT-INT"))
print(pwlf.slopes("PoT-SIGN"))
print(pwlf.slopes("APoT"))
print(pwlf.slopes("APoT-INT"))
print(pwlf.slopes("APoT-SIGN"))
```

### Evaluate PWLF, PoT-PWLF, and APoT-PWLF approximated Sigmoid

```python
xin = np.array([0.0, 12.3, 55.5, 100.0])
yout = pwlf(xin, "original")
print(yout)
yout = pwlf(xin, "PoT")
print(yout)
yout = pwlf(xin, "APoT")
print(yout)
```
